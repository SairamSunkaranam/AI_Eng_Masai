```python
with open("notes.txt","w") as f:
  f.write("Hello Students\n")
  f.write("ML is good")
```

```python
with open("notes.txt","r") as f:
  data = f.read()
print(data)
```

```python
#append
with open("notes.txt","a") as f:
  f.write("\nML is good")
```

```python
with open("notes.txt","r") as f:
  data = f.read()
print(data)
print(type(data))
```

```python
#readline()
with open("notes.txt","r") as f:
  data = f.readline()
print(data)
print(type(data))
```

```python
#read lines
with open("notes.txt","r") as f:
  data = f.readlines()
print(data)
print(type(data))
```

```python
#csv

import csv

with open("marks.csv","w",newline="") as f:
  writer = csv.writer(f)
  writer.writerow(["name","marks"])
  writer.writerow(["Sam",85])
  writer.writerow(["Tam",90])
```

```python
import csv

with open("marks.csv","r") as f:
 reader = csv.reader(f)
 for row in reader:
  print(row)

```

```python
#DictReader()

import csv

with open("quickbite_orders.csv","r") as f:
  reader = csv.DictReader(f)
  menu = list(reader)
  #print(menu)

  for item in menu:
    print(item["item_name"])
```
