Data Strcutures

List, Tuple, Dict and Set

```python
#List
students = ["Sandhya", "Rahul", "Anjali"]
print(students)
```

```python
print("Student is: ", students[2])
print("Student is: ", students[-1])
```

```python
# add, update, remove to the list
# list are mutable - can be modified
students.append("Meena")
print(students)
```

```python
#lenght of list
print(len(students))
```

```python
# Update

students[1] = "Ravi"
print(students)
```

```python
# Remove
removed = students.pop(1)
print("Removed Student: ",removed)
print("Remaining:",students)
```
