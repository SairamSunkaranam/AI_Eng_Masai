Data Strcutures

List, Tuple, Dict and Set

```python
#List
students = ["Sandhya", "Rahul", "Anjali"]
print(students)
```

```python
print("Student is: ", students[2])
print("Student is: ", students[-1])
```

```python
# add, update, remove to the list
# list are mutable - can be modified
students.append("Meena")
print(students)
```

```python
#lenght of list
print(len(students))
```

```python
# Update

students[1] = "Ravi"
print(students)
```

```python
# Remove
removed = students.pop(1)
print("Removed Student: ",removed)
print("Remaining:",students)
```

#List allows duplicates, Mutable, ordered and can contain any data types

# Tupes - Uses normal braces

```python
location = (12.3850, 99.4456)
print(location)
```

```python
print(location[1])
```

```python
print(len(location))
```

```python
#Immutable - Can't change the values
location[0] = 11.12
print(location)
```

```python
#Assigning completely new value to tuple is allowed
location = (50.5, 150.5)
print(location)
```

```python
# Tuple allows duplicates
location = (17.7, 133.3, 133.3)
print(location)
```

#Tuple is immultable, allowes duplicates and ordered

# Dictionary: Key-Vaule pairs and used curly brances {}

```python
st = {"name":"Sam","roll_number":1,"course":"Python"}
print(st)
```

```python
print(st["name"])
```

```python
print(st.values())
print(st.keys())
```

```python
#Each Key-Vaule pair in the dict called as item
print(st.items())
```

```python
print(st.get("name"))
```

```python
st.update({"city":"Hyderabad"})
print(st)
```

```python
removed = st.pop("name")
print(st)
```

# Dict: Mutable, Ordered, Duplicate Key (Not accepted), allowes duplicate values

```python
st.update({"city":"Chennai"})
print(st)
```

```python
st["course"] = "AI"
print(st)
```

```python
marks = {
    "Maths": 80,
    "Maths": 90
}
print(marks)
```

# SET

```python
attendance = ["Sam", "pam", "ram", "tam", "ram"]
print(attendance)
```

```python
unique_students = set(attendance)
print(unique_students)
```

```python
fruits = {"Apple","Banana","Mango","Apple"}
print(fruits)
```

```python
print(sorted(fruits))
print(len(fruits))
```

```python
fruits.add("orange")
print(fruits)
```

```python
# Values should be given in the SET to refer. Indexex can't be provided
fruits.remove("Banana")
print(fruits)
```

```python
# discard won't be returning an error even you provided non-existant value. But the remove function does
fruits.discard("Grapes")
fruits.remove("Grapes")
```

# SET mutable, no index, won't allow duplicates, no order
