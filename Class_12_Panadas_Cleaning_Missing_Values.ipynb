```python
after_drop = df.dropna()
after_drop.head()
```

```python
def report_loss(before, after):
    lost = len(before) - len(after)
    pct = lost / len(before) * 100
    print(f"{len(before)} rows -> {len(after)} rows | lost {lost} ({pct:.1f}%)")

after_drop = df.dropna()
report_loss(df, after_drop)
```

```python
rows_after = df.dropna(subset=['delivery_min', 'rating'])
report_loss(df, rows_after)
print(rows_after)
```

#Filling Instead of dropping

```python
print('real average rating :', round(df['rating'].mean(), 2))
print('after filling with 0 :', round(df['rating'].fillna(0).mean(), 2))
```

```python
print('mean : ', round(df['delivery_min'].mean(), 1))
print('median: ', df['delivery_min'].median())
```

```python
print('mean  : ', round(df['amount'].mean(), 1))
print('median: ', df['amount'].median())
```

```python
df['delivery_min'] = df['delivery_min'].fillna(round(df['delivery_min'].mean()))

df['rating'] = df['rating'].fillna(round(df['rating'].mean()))
```

```python
df.isna().sum()
```

```python
print('mean : ', round(df['delivery_min'].mean(), 1))
print('median: ', df['delivery_min'].median())

print('mean  : ', round(df['rating'].mean(), 1))
print('median: ', df['rating'].median())
```
