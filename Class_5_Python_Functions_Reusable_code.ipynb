```python
# return vs print
def add(a,b):
  print(a+b)
add(10,20)
```

```python
def add(a,b):
  return a+b

result = add(10, 20)
print(result)
```

```python
def add_with_print(a,b):
  print(a+b)
def add_with_return(a,b):
  return a+b
result_print = add_with_print(10,20)
print(result_print)
result_return = add_with_return(10,20)
print(result_return)
```

```python
#Positional arguments vs Keyword arguments

def final_price(price, discount):
  return price - discount
```

```python
print(final_price(500, 50)) #Positional Arguments
print(final_price(discount=100,price=1000)) #Keyword Aruguments
```

```python
# Rule of thumb - Guidlines
# Positional argument = position matters
# key word = Name matters
```

```python
def delivery_fee(amount, fee=40):
  if amount >= 300:
    return 0
  return fee

print(delivery_fee(390))
print(delivery_fee(180, 70))
```

```python
# Mini Project

def process_order(item, quantity, price):

  # 1. Clean item name
  item = " ".join(item.split()).title()

  # 2. Calculate subtotal
  sub_total = quantity * price

  # 3. Devlivery fee
  if sub_total >= 300:
    delivery_fee = 0
  else:
    delivery_fee = 40

  # Calculate the total
  total = sub_total + delivery_fee

  #Display
  print("Item:", item)
  print("Quantity:", quantity)
  print("Subtotal:", sub_total)
  print("Delivery Fee:", delivery_fee)
  print("Total:", total)


process_order("    Panner     Tikka Masala       ", 2, 160)
```
