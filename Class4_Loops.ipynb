```python
#Break and Continue
#break = entire loop will stop
#continue = current iteraton skip, it will go to the next iteration

for number in range(1,6):
  if number == 3:
    break
  print(number)
```

```python
for number in range(1,6):
  if number == 3:
    continue
  print(number)
```

```python
# mini project
raw_orders = [
    {"item_name":  "Pizza", "quantity": 2, "price": 10.99},
    {"item_name":  "Burger", "quantity": 1, "price": 8.99},
    {"item_name":  "Salad", "quantity": 3, "price": 6.99},
    {"item_name":  "Pasta", "quantity": 1, "price": 12.99}
]

print(raw_orders)
```

```python
# Step 2 Process Every Order
processed_orders = []
for order in raw_orders:
  name = " ".join(order["item_name"].split()).title() #order["item_name"]
  #print(" ".join(order["item_name"].split()))
  #subtotal = round((order["quantity"] * order["price"]),2)
  subtotal = order["quantity"] * order["price"]
  print(subtotal)
  if subtotal >= 300:
    delivery_fee = 0
  else:
    delivery_fee = 30
  final_total = subtotal + delivery_fee
  print(final_total)

  processed_orders.append({
      "item_name": name,
      "quantity": order["quantity"],
      "total": final_total,
      "delivery_fee": delivery_fee
  })

print(processed_orders)
```

```python
total_revenue = 0

for order in processed_orders:
  total_revenue += order["total"]

print(total_revenue)
```
