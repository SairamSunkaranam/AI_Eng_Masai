```python
import pandas as pd
import numpy as np

try:
    orders = pd.read_csv('orders_clean.csv')
except FileNotFoundError:
    # Re-create the clean file from last session
    orders = pd.read_csv('orders.csv', na_values={'delivery_min': [999]})
    orders['delivery_min'] = orders['delivery_min'].fillna(orders['delivery_min'].median()).astype(int)
    orders['rating'] = orders['rating'].fillna(orders['rating'].median()).astype(int)
    orders.to_csv('orders_clean.csv', index=False)

restaurants = pd.read_csv('restaurants.csv')
customers = pd.read_csv('customers.csv')

print(orders.shape, restaurants.shape, customers.shape)
```

#Merging tables = SQL JOIN

```python
left = orders.merge(restaurants, on='restaurant_id', how='left')
left.shape
```

```python
left
```

```python
right = orders.merge(restaurants, on='restaurant_id', how='right')
right.shape
```

```python
outer = orders.merge(restaurants, on='restaurant_id', how='outer', indicator=True)
outer.shape
```

```python
summary = left.groupby('name').agg(
    orders = ('order_id', 'count'),
    revenue = ('amount', 'sum'),
    avg_delivery = ('delivery_min', 'mean'),
    best_rating = ('rating', 'max')
).round(1)

summary.sort_values('revenue', ascending=False)
```
